```python
import spacy

nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")
nlp.max_length = 7000000
doc = nlp(text)

sentences = [[token.text for token in sent] for sent in doc.sents]
```

```python
sentences
```

```python

trigram=dict()
bigram=dict()
for sentence in sentences:
    new_sentence = ["<s>", "<s>"] + sentence + ["</s>"]
    for i in range(len(new_sentence)-2):
        w1, w2, w3 = new_sentence[i],new_sentence[i+1],new_sentence[i+2]
        if (w1,w2) in trigram:
            bigram[(w1,w2)] += 1
            if w3 in trigram[(w1,w2)]:
                trigram[(w1,w2)][w3] += 1
            else:
                trigram[(w1,w2)][w3] = 1
        else:
            bigram[(w1,w2)] = 1
            trigram[(w1,w2)] = {w3 : 1}
```

```python
def get_prob(w1, w2, w3):
    return trigram[(w1, w2)][w3] / bigram[(w1, w2)]
```

```python
vocab = set(words)
V = len(vocab)
```

```python
def laplace_prob(w1, w2, w3):

    count_w1_w2_w3 = trigram.get((w1, w2), {}).get(w3, 0)

    count_w1_w2 = bigram.get((w1, w2), 0)

    numerator = count_w1_w2_w3 + 1
    denominator = count_w1_w2 + V # V is already the length of vocab

    if denominator == 0:
        return 0.0

    return numerator / denominator
```

```python
import random

def generate(max_len=20):
    w1, w2 = "<s>", "<s>"
    sentence = []

    for _ in range(max_len):
        probs = []
        words = []

        candidates = trigram.get((w1, w2), {})

        words = list(candidates.keys())
        probs = [laplace_prob(w1, w2, w) for w in words]

        k = 5

        pairs = list(candidates.items())
        pairs.sort(key=lambda x: x[1], reverse=True)

        top_words = [w for w, _ in pairs[:k]]
        top_probs = [laplace_prob(w1, w2, w) for w in top_words]

        next_word = random.choices(top_words, weights=top_probs)[0]

        if next_word == "</s>":
            break

        sentence.append(next_word)
        w1, w2 = w2, next_word

    return " ".join(sentence)
```
